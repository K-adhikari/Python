```python
grades = [100, 100, 90, 40, 80, 100, 85, 70, 90, 65, 90, 85, 50.5]


# Write a function to sum all grades

def grades_sum(scores):
  sum = 0
  for score in scores:
    sum += score
  return sum

#grades_sum(grades)


# Write a function to get average grades. Make sure to call the grades_sum function here

def grades_average(grades_input):
  total_grade = grades_sum(grades_input)
  average = total_grade / float(len(grades_input))
  return average
  

#grades_average(grades)


# Compute variance in grades. Make sure to call the grades_average function here

def grades_variance(scores):
  average = grades_average(scores)
  variance = 0
  for score in scores:
    squared_difference = (average - score) ** 2
    variance += squared_difference
  return variance / float(len(scores))

#grades_variance(grades)


# Define a function, grades_std_deviation, that takes one argument called variance.

def grades_std_deviation(variance):
  
  return variance ** 0.5

variance = grades_variance(grades)  

#grades_std_deviation(variance)


print("Total grades: ", grades_sum(grades))
print("Grade average: ",  grades_average(grades))
print("Grade variance: ",  grades_variance(grades))
print("Grade Standard devaition: ", grades_std_deviation(variance))
```
