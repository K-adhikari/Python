1. Define a function called reverse that takes a string text and returns that string in reverse.

```python
def reverse(text):
    reversed_text = ""
    l = len(text) - 1
    while l >= 0:
        reversed_text = reversed_text + text[l]
        l -= 1
    return reversed_text
  
    
# Usage:
reverse("Hello World")
```

2. Define a function called anti_vowel that takes one string, text, as input and returns the text with all of the vowels removed.

```python
def anti_vowel(text):
  new_text = ""
  vowels = {"a", "e", "i" , "o" , "u", "A", "E", "I", "O", "U"}
  for i in text:
    if i not in vowels:
      new_text += i
  return new_text


#Usage:
anti_vowel("Hey You!")
```

3. Define a function scrabble_score that takes a string word as input and returns the equivalent scrabble score for that word.

```python
score = {"a": 1, "c": 3, "b": 3, "e": 1, "d": 2, "g": 2, 
         "f": 4, "i": 1, "h": 4, "k": 5, "j": 8, "m": 3, 
         "l": 1, "o": 1, "n": 1, "q": 10, "p": 3, "s": 1, 
         "r": 1, "u": 1, "t": 1, "w": 4, "v": 4, "y": 4, 
         "x": 8, "z": 10}

def scrabble_score(word):
  word = word.lower()
  game_score = 0
  for char in word:
    for key in score:
      if char == key:
        game_score += score[key]
  return game_score


#Usage:
scrabble_score("Pizzeria")
```

4. Write a function called censor that takes two strings, text and word, as input. It should return the text with the word you chose replaced with asterisks.

```python
def censor(text, word):
  replacement = '*' * len(word)
  new_text = text.replace(word, replacement)
  return new_text


#Usage:
censor("this hack is wack hack", "hack")
```

5. Define a function called count that has two arguments called sequence and item. Return the number of times the item occurs in the list.

```python
def count(sequence, item):
  count = 0
  for i in sequence:
    if i == item:
      count += 1
  return count


#Usage:
count([1, 2, 1, 1], 1) 
```

6. Define a function called purify that takes in a list of numbers, removes all odd numbers in the list, and returns the result.

```python
def purify(sequence):
  new_sequence = []
  for i in sequence:
    if i%2 == 0:
      new_sequence.append(i)
  return new_sequence


#Usage:
purify([1,2,3,4,6,7,8])
```

7. Define a function called product that takes a list of integers as input and returns the product of all of the elements in the list.

```python
def product(list):
  new_product = 1
  for item in list:
    new_product = new_product * item
  return new_product


#Usage: 
product([4,5,5])
```

8. Write a function remove_duplicates that takes in a list and removes elements of the list that are the same.

```python
def remove_duplicates(list):
  new_list = []
  for item in list:
    if item not in new_list:
      new_list.append(item)
  return new_list


#Usage:
remove_duplicates([1, 1, 2, 2, 2, 3, 3, 5])
```

9. write a function to find the median of a list.

```python
def median(list):
  sorted_list = sorted(list)
  if len(sorted_list) % 2 == 1:
    position = (len(sorted_list) - 1) // 2
    median = sorted_list[position]

  else:
    position = len(sorted_list) // 2
    median = (sorted_list[position] + sorted_list[position - 1]) / 2.0

  return median


#Usage:
median([1, 2, 3, 4, 5]) # Gives 3
median([0, 1, 2, 3, 4, 5])   # Gives 2.5
```
