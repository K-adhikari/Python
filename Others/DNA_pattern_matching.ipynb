1. Write a function to count patterns in a DNA sequence.

```python
 def PatternCount(Text, Pattern):
    
    count = 0
    for i in range(len(Text)-len(Pattern)+1):
        if Text[i:i+len(Pattern)] == Pattern:
            count = count+1
    return count


# Usage example: 

Text = "ATCAATGATCAACGTAAGCTTCTAAGCATGATCAAGGTGCTCACACAGTTTATCCACAACCTGAGTGGATGACATCAAGATAGGTCGTT\
        GTATCTCCTTCCTCTCGTACTCTCATGACCACGGAAAGATGATCAAGAGAGGATGATTTCTTGGCCATATCGCAATGAATACTTGTGAC\
        TTGTGCTTCCAATTGACATCTTCAGCGCCATATTGCGCTGGCCAAGGTGACGGAGCGGGATTACGAAAGCATGATCATGGCTGTTGTTC\
        GTTTATCTTGTTTTGACTGAGACTTGTTAGGATAGACGGTTTTTCATCACTGACTAGCCAAAGCCTTACTCTGCCTGACATCGACCGTA\
        AATTGATAATGAATTTACATGCTTCCGCGACGATTTACCTCTTGATCATCGATCCGATTGAAGATCTTCAATTGTTAATTCTCTTGCCT\
        CGACTCATAGCCATGATGAGCTCTTGATCATGTTTCCTTAACCCTCTATTTTTTACGGAAGAATGATCAAGCTGCTGCTCTTGATCATCGTTTC"
Pattern = "TGATCA"

PatternCount(Text, Pattern)
```

2. Find the frequency of K-mer of desired length in the sequence.

```python
def FrequencyMap(Text, k):
    freq = {}
    n = len(Text)
    for i in range(n-k+1):
        Pattern = Text[i:i+k]
        freq[Pattern] = 0
        for j in range(n-k+1):
            if Text[j:j+k] == Pattern:
                freq[Pattern] += 1
    
    return freq

# Usage example:

Text = "CGATATATCCATAG"
k = 3

FrequencyMap(Text, k)
```

3. Find the most Frequent K-mer.

```python
def FrequentWords(Text, k):
    words = []
    freq = FrequencyMap(Text, k)
    m = max(freq.values())
    for key in freq:
        if freq[key] == m:
            words.append(key)
    return words

# Usage example:

Text = "ACGTTGCATGTCGCATGATGCATGAGAGCT"
k = 4

FrequentWords(Text, k)
```

4. Find the reverse of a sequence.

```python
def Reverse(Pattern):
    rev = ""
    for char in Pattern:
        rev = char + rev
        
    return rev

# Usage example:

Reverse("AAAACCCGGT")
```

5. Find the complement of a given sequence.

```python
def Complement(Pattern):
    Complement = ""
    for char in Pattern:
        if char == 'A':
            Complement = Complement + "T"
        if char == 'T':
            Complement = Complement + "A"
        if char == 'G':
            Complement = Complement + "C"
        if char == 'C':
            Complement = Complement + "G"
    return Complement

# Usage example:

Complement("AAAACCCGGT")
```

6. Find the reverse complement of a given sequence.

```python
def ReverseComplement(Pattern):   
    Rev = Reverse(Pattern)
    Rev_com = Complement(Rev)
    return Rev_com

# Usage example:

ReverseComplement("AAAACCCGGT")
```

7. Find positions in a sequence matching a pattern.

```python
def PatternMatching(Pattern, Genome):
    positions = [] # output variable
    for i in range(len(Genome)-len(Pattern)+1):
        if Genome[i:i+len(Pattern)] == Pattern:
            positions.append(i)
    return positions

# Usage example:

PatternMatching('ATAT', 'GATATATGCATATACTT')
```
