1. Use list slicing to print out every odd element of my_list from start to finish.

```python
my_list = range(1, 11)

print(list(my_list[::2]))
```

2. Create a variable called backwards and set it equal to the reversed version of my_list.

```python
backwards = my_list[::-1]

print(list(backwards))
```

3. Create a variable, backwards_by_tens, and set it equal to the result of going backwards through to_one_hundred by tens.

```python
to_one_hundred = range(101)

backwards_by_tens = to_one_hundred[::-10]

print(list(backwards_by_tens))
```

4. Create a list, squares, that consists of the squares of the numbers 1 to 10.
Use filter() and a lambda expression to print out only the squares that are between 30 and 70 (inclusive).

```python
squares = [1, 4, 9, 16, 25, 36, 49, 64, 81, 100]

print(list(filter(lambda x: 30 <= x <= 70, squares)))
```

6. Use a list comprehension to create a list, threes_and_fives, that consists only of the numbers between 1 and 15 (inclusive) that are evenly divisible by 3 or 5.

```python
threes_and_fives = [x for x in range(1, 16) if x % 3 == 0 or x % 5 == 0]

print(threes_and_fives)
```

7. The string in the editor is garbled in two ways:

Our message is backwards.
The letter we want is every other letter.
Use list slicing to extract the message and save it to a variable called message.

```python
garbled = "!XeXgXaXsXsXeXmX XtXeXrXcXeXsX XeXhXtX XmXaX XI"

reversed_garbled = garbled[::-1]
message = reversed_garbled[::2]
message
```

8. Create a new variable called message. Set it to the result of calling filter() with the appropriate lambda that will filter out the "X"s. The second argument will be garbled. Finally, print your message.

```python
garbled = "IXXX aXXmX aXXXnXoXXXXXtXhXeXXXXrX sXXXXeXcXXXrXeXt mXXeXsXXXsXaXXXXXXgXeX!XX"
message = filter(lambda x: x != 'X', garbled)
print(''.join(message))
```
